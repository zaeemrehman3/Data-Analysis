# Vectorized Operations

```python
lst1 = [1,3,5,7,9]
lst2 = [2,4,6,8,10]
```

```python
lst3 = []
for n in zip(lst1,lst2):
    lst3.append(sum(n))
lst3
```

```python
list(zip(lst1,lst2))
```
