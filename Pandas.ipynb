#### adding new column

```python
empD['tax%'] = 15
empD
```

```python
def decide_tax_percentage(sal):
    if sal >= 200000:
        return 20
    elif sal >= 150000:
        return 15
    elif sal >= 100000:
        return 10
    else:
        return 5
```

```python
empD['tax%'] = empD.Salary.apply(decide_tax_percentage)
```

```python
empD
```

```python
empD.Salary.apply(lambda x:20 if x>=75000 else 15 if x>=50000 else 10 if x>=25000 else 5)
```

```python
empD['tax%'] = empD.Salary.apply(lambda x:20 if x>=75000 else 15 if x>=50000 else 10 if x>=25000 else 5)
```

```python
empD
```

```python
empD['tax_amount'] = empD.Salary.apply(lambda x: x*.20 if x>=75000 else x*.15 if x>=50000 else x*.10 if x>=25000 else x*.05)
```

```python
empD
```

```python
empD['net'] = empD['Salary'] - empD['tax_amount']
empD
```
